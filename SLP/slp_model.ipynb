```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#Numerical Example

```python
x=np.array([1,2])
y=1
w=np.array([0.2,-0.5])
b=0.1
eta=0.1
```

```python
for i in range(1,4):
  print(f"Epoch {i}:")
  z=np.dot(w,x) + b
  y_hat=0 if z<=0 else 1
  error=y-y_hat
  w=w+(eta*error*x)
  b=b+(eta*error)
  print(f"   Weighted Sum = {z:.2f}")
  print(f"   Prediction = {y_hat}")
  print(f"   Error = {error}")
  print(f"   Weights = {w}")
  print(f"   Bias = {b:.2f}\n")
```

```python
x=np.array([2,1])
z=np.dot(w,x)+b
y_hat=0 if z<=0 else 1
print("Prediction = ",y_hat)
```

##Without Bias

```python
x = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[1]])
w = np.array([0.1,0.3])
eta = 0.2
theta = 0.5
```

```python
x[0][0]
```

```python
for j in range(4):
  w11=[]
  w21=[]
  pred=[]
  error=[]

  for i in range(len(x)):
    z = np.dot(x[i],w)
    y_hat = 0 if z<0.5 else 1
    w11.append(w[0])
    w21.append(w[1])
    err=y[i]-y_hat
    pred.append(y_hat)
    error.append(err)
    w=w+(eta*err*x[i])

  data = {"x1":x[:,0],"x2":x[:,1],"W11":w11,"W21":w21,"ŷ":pred,"y":y.flatten(),"Error":error}
  df=pd.DataFrame(data)
  print(f"Epaoch {j+1}")
  display(df)
  print()
```

```python
w  #After all Epochs
```

```python
x1 = np.linspace(-1, 2, 100)
x2 = (0.5-w[0]* x1) /w[1]

# plot points
plt.scatter(x[:,0], x[:,1], c=y.flatten(), cmap='bwr')
plt.plot(x1, x2, label="Without Bias")

plt.legend()
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Perceptron Decision Boundary")
plt.axis("Equal")
plt.show()
```

##With Bias

```python
x = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[1]])
w = np.array([0.1,0.3])
eta = 0.2
theta = 0.5
```

```python
b=[0.2]

for j in range(4):
  w11=[]
  w21=[]
  bs=[]
  pred=[]
  error=[]

  for i in range(len(x)):
    z = np.dot(x[i],w)+b
    y_hat = 0 if z<0.5 else 1
    w11.append(w[0])
    w21.append(w[1])
    bs.append(b[0])
    err=y[i]-y_hat
    pred.append(y_hat)
    error.append(err)
    w=w+(eta*err*x[i])
    b=b+(eta*err)

  data = {"x1":x[:,0],"x2":x[:,1],"W11":w11,"W21":w21,"b":bs,"ŷ":pred,"y":y.flatten(),"Error":error}
  df=pd.DataFrame(data)
  print(f"Epoch {j+1}")
  display(df)
  print()
```

```python
print(f"After Final Epoch:\n   w: {w}\n   b:{b}")
```

```python
plt.scatter(x[:,:1],x[:,1:],c=y.flatten(),cmap='bwr')

x1=np.linspace(-1,2,100)
x2=(0.5-b-w[0]*x1)/w[1]
plt.plot(x1,x2,label="With Bias")
plt.legend()
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Perceptron Decision Boundary")
plt.show()
```

**Explanation of the Decision Boundary Plot**

The plotted line represents the **decision boundary** learned by the perceptron.
The decision boundary is defined by the equation:

[
$w_1x_1 + w_2x_2 + b = 0.5$
]

Any point for which:

[
$w_1x_1 + w_2x_2 + b \ge 0.5$
]

is classified as **class 1**, and any point for which:

[
$w_1x_1 + w_2x_2 + b < 0.5$
]

is classified as **class 0**.

**Boundary separates the Classes**
